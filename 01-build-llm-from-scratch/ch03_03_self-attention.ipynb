```python
import torch

inputs = torch.tensor([
    [0.43, 0.15, 0.89], # Your      (x^1)
    [0.55, 0.87, 0.66], # journey   (x^2)
    [0.57, 0.85, 0.64], # starts    (x^3)
    [0.22, 0.58, 0.33], # with      (x^4)
    [0.77, 0.25, 0.10], # one       (x^5)
    [0.05, 0.89, 0.55], # step      (x^6)
])
```

```python
query = inputs[1]

attn_scores_2 = torch.empty(inputs.shape[0])
# print(inputs.shape) # torch.Size([6, 3])
# print(attn_scores_2) # tensor([9.6668e+21, 1.1771e-42, 0.0000e+00, 0.0000e+00, 0.0000e+00, 0.0000e+00])
for i, x_i in enumerate(inputs):
    attn_scores_2[i] = torch.dot(x_i, query)

print(attn_scores_2) # tensor([0.9544, 1.4950, 1.4754, 0.8434, 0.7070, 1.1648])
```

```python
res = 0
for idx, element in enumerate(inputs[0]):
    res += inputs[0][idx] * query[idx]

print(res)
print(torch.dot(inputs[0], query))
```

```python
print(attn_scores_2) # tensor([0.9544, 1.4950, 1.4754, 0.8434, 0.7070, 1.1648])

attn_weights_2_tmp = attn_scores_2 / attn_scores_2.sum()
print(f"어텐션 가중치: {attn_weights_2_tmp}")
print(f"합: {attn_weights_2_tmp.sum()}")
```

```python
def softmax_naive(x):
    return torch.exp(x) / torch.exp(x).sum(dim=0)

attn_weights_2_naive = softmax_naive(attn_scores_2)
print(f"어텐션 가중치: {attn_weights_2_naive}")
print(f"합: {attn_weights_2_naive.sum()}")
```

```python
attn_weights_2 = torch.softmax(attn_scores_2, dim=0)
print(f"어텐션 가중치: {attn_weights_2}")
print(f"합: {attn_weights_2.sum()}")
```

```python
print(query) # inputs[1]

context_vec_2 = torch.zeros(query.shape)
print(context_vec_2, context_vec_2.shape)

for i, x_i in enumerate(inputs):
    context_vec_2 += attn_weights_2[i] * x_i

print(context_vec_2)
```

```python
# 모든 문맥 벡터를 계산

attn_scores = torch.empty(6, 6)

# print(inputs)
for i , x_i in enumerate(inputs):
    for j, x_j in enumerate(inputs):
        attn_scores[i, j] = torch.dot(x_i, x_j)

print(f"attention scores:\n{attn_scores}")
```

```python
# 행렬 곱셈을 사용

print(f"inputs:\n{inputs}\n")
# inputs = torch.tensor([
#     [0.43, 0.15, 0.89], # Your      (x^1)
#     [0.55, 0.87, 0.66], # journey   (x^2)
#     [0.57, 0.85, 0.64], # starts    (x^3)
#     [0.22, 0.58, 0.33], # with      (x^4)
#     [0.77, 0.25, 0.10], # one       (x^5)
#     [0.05, 0.89, 0.55], # step      (x^6)
# ])

attn_scores = inputs @ inputs.T
print(f"attn_scores:\n{attn_scores}\n")

attn_weights = torch.softmax(attn_scores, dim=-1)
print(f"attn_weights:\n{attn_weights}")
```

```python
print(f"모든 행의 합: {attn_weights.sum(dim=1)}")
```

```python
all_context_vecs = attn_weights @ inputs
print(all_context_vecs)
print(context_vec_2)
```
