```python
from importlib.metadata import version
import tiktoken
print("tiktoekn version: ", version("tiktoken"))
```

```python
tokenizer = tiktoken.get_encoding("gpt2")
```

```python
text = (
    "Hello, do you like tea? <|endoftext|> In the sunlit terraces"
    " of soumeunkownPlace."
)

integers = tokenizer.encode(text, allowed_special={"<|endoftext|>"})
print(integers)

strings = tokenizer.decode(integers)
print(strings)
```

#### 연습문제 2.1
### 알지 못하는 단어에 대한 바이트 페어 인코딩
- tiktoken 라이브러리의 BPE 토크나이저를 알지 못하는 단어 'Akwir ier"에 적용해서 개별 토큰 ID를 출력해보기
- 출력 리스트의 각 정수 값에 decode 메서드를 호출하여 매핑을 만들어 보기
- decode 메서드를 토큰 ID 리스트에 적용하여 원본 입력을 재구성할 수 있는지 확인

```python
tokenizer_list = tiktoken.list_encoding_names()
print(tokenizer_list)
# ['gpt2', 'r50k_base', 'p50k_base', 'p50k_edit', 'cl100k_base', 'o200k_base', 'o200k_harmony']

tokenizer = tiktoken.get_encoding("gpt2")

text = "Akwir ier"
integers = tokenizer.encode(text)
print(integers) # [33901, 86, 343, 220, 959]
print()

for i in integers:
    print(f"{tokenizer.decode([i])} -> {i}")

strings = tokenizer.decode(integers)
print(strings)
```
