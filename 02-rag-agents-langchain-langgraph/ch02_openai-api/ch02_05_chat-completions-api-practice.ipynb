```python
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv("../.env")
client = OpenAI()
```

### API 호출

```python
# Chat Completions API 호출하기
response = client.chat.completions.create(
    model="gpt-6-luna",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "안녕하세요! 저는 존이라고 합니다!"},
    ],
)
print(response.to_json(indent=2))
```

```python
# Responses API 호출하기
response = client.responses.create(
    model="gpt-6-luna",
    input=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "안녕하세요! 저는 존이라고 합니다!"}
    ]
)
print(response.to_json(indent=2))
```

### 스트리밍으로 응답 얻기

```python
# Chat Completions API
response = client.chat.completions.create(
    model="gpt-6-luna",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "안녕하세요! 저는 존이라고 합니다!"}
    ],
    stream=True,
)

for chunk in response:
    content = chunk.choices[0].delta.content
    if content is not None:
        print(content, end="", flush=True)
```

```python
# Responses API
response = client.responses.create(
    model="gpt-6-luna",
    instructions="You are a helpful assistant.",
    input="안녕하세요! 저는 존이라고 합니다!",
    stream=True,
)

for event in response:
    if event.type == "response.output_text.delta":
        print(event.delta, end="", flush=True)
```

### JSON 모드

```python
# Chat Compeltions API
response = client.chat.completions.create(
    model="gpt-6-luna",
    messages=[
        {
            "role": "system",
            "content": '인물 목록을 다음 JSON 형식으로 출력해 주세요.\n{"people": ["aaa", "bbb]}',
        },
        {
            "role": "user",
            "content": "옛날 옛적에 할아버지와 할머니가 살고 있었습니다."
        }
    ],
    response_format={"type": "json_object"}
)
print(response.choices[0].message.content)
```

```python
# Response API
response = client.responses.create(
    model="gpt-6-luna",
    instructions='인물 목록을 다음 JSON 형식으로 출력해 주세요.\n{"people": ["aaa", "bbb"]}',
    input=(
        "옛날 옛적에 할아버지와 할머니가 살고 있었습니다.\n"
        "등장인물을 JSON으로 출력해 주세요."
    ),
    text={"format": {"type": "json_object"}}
)
print(response.output_text)
```

### Vision(이미지 입력)

```python
# Chat Completions API
image_url = "https://raw.githubusercontent.com/ychoi-kr/langchain-book/main/cover.jpg"

response = client.chat.completions.create(
    model="gpt-6-luna",
    messages=[
        {
            "role": "user",
            "content": [
                {"type": "text", "text": "이미지를 설명해 주세요."},
                {"type": "image_url", "image_url": {"url": image_url}},
            ]
        }
    ]
)
print(response.choices[0].message.content)
```

```python
# Responses API
image_url = "https://raw.githubusercontent.com/ychoi-kr/langchain-book/main/cover.jpg"

response = client.responses.create(
    model="gpt-6-luna",
    input=[
        {
            "role": "user",
            "content": [
                {"type": "input_text", "text": "이미지를 설명해 주세요"},
                {"type": "input_image", "image_url": image_url, "detail": "auto"},
            ]
        }
    ]
)
print(response.output_text)
```
