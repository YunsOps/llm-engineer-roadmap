````python
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv("../.env")

client = OpenAI()
````

````python
prompt = '''\
다음 요리의 레시피를 생각해 주세요.

요리명: """
{dish}
"""
'''

def generate_recipe(dish: str) -> str:
    response = client.chat.completions.create(
        model="gpt-6-luna",
        messages=[
            {"role": "user", "content": prompt.format(dish=dish)}
        ]
    )
    return response.choices[0].message.content
````

````python
def generate_recipe(dish: str) -> str:
    response = client.chat.completions.create(
        model="gpt-6-luna",
        messages=[
            {"role": "system", "content": "사용자가 입력한 요리의 레시피를 생각해 주세요."},
            {"role": "user", "content": f"{dish}"}
        ]
    )
    return response.choices[0].message.content
````

````python
recipe = generate_recipe("카레")
print(recipe)
````

````python
# 프롬프트에서 출력 형식을 지정하는 경우도 많다.
system_prompt = """\
사용자가 입력한 요리의 레시피를 만들어 주세요.

출력은 다음 JSON 형식으로 해주세요.

```
{
 "재료": ["재로1", "재료2"]
 "순서": ["순서1", "순서2"]
}
```
"""

response = client.chat.completions.create(
    model="gpt-6-luna",
    messages=[
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": "카레"}
    ]
)
print(response.choices[0].message.content)
````
