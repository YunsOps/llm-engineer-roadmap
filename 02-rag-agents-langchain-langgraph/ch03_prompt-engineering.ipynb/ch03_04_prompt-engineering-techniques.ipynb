```python
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv("../.env")

client = OpenAI()
```

### Zero-shot 프롬프팅

```python
# 입력 텍스트가 긍정적인지 부정적인지 판단하는 일명 감정 분석을 위한 프롬프트의 예는 다음과 같다.
response = client.chat.completions.create(
    model="gpt-6-luna",
    messages=[
        {
            "role": "system",
            "content": "입력을 긍정적, 부정적, 중립 중 하나로 분류해 주세요."
        },
        {
            "role": "user",
            "content": "ChatGPT는 프로그래밍 고민을 많이 해결해 줍니다."
        }
    ]
)
print(response.choices[0].message.content)
```

### Few-shot 프롬프팅

```python
# 먼저 Zero-shot 프롬프팅으로 “입력이 AI와 관련이 있는지 답변해 주세요.”라고 지시해 보겠다.
response = client.responses.create(
    model="gpt-6-luna",
    input=[
        {"role": "system", "content": "입력이 AI와 관련이 있는지 답변해 주세요."},
        {"role": "user", "content": "ChatGPT는 미우 편리하다"}
    ]
)
print(response.output_text)
```

```python
# 답변을 원하는 텍스트 앞에 몇 가지 시연 예시를 포함한 프롬프트를 작성
response = client.responses.create(
    model="gpt-6-luna",
    input=[
        {"role": "system", "content": "입력이 AI와 관련이 있는지 답변해 주세요."},
        {"role": "user", "content": "AI의 진화는 놀랍다."},
        {"role": "assistant", "content": "true"},
        {"role": "user", "content": "오늘은 날씨가 좋다"},
        {"role": "assistant", "content": "false"},
        {"role": "user", "content": "ChatGPT는 매우 편리하다"},
    ]
)
print(response.output_text)
```

### Few-shot 프롬프팅의 다른 형식

```python
prompt="""\
입력이 AI와 관련이 있는지 답변해 주세요.

Q: AI의 진화는 놀랍다
A: true
Q: 오늘은 날씨가 좋다
A: false
Q: ChatGPT는 매우 편리하다
A:
"""

response = client.responses.create(
    model="gpt-6-luna",
    input=prompt,
)
print(response.output_text)
```

```python
# Few-shot 예시가 대화 이력이 아님을 강조하고 싶다면, 다음과 같이 system 메시지의 “name”을 “example_user”나 “example_assistant”로 설정하면 된다.
response = client.chat.completions.create(
    model="gpt-6-luna",
    messages=[
        {"role": "system", "content": "입력이 AI와 관련이 있는지 답변해 주세요."},
        {"role": "system", "name": "example_user", "content": "AI의 진화는 놀랍다"},
        {"role": "system", "name": "example_assistant", "content": "true"},
        {"role": "system", "name": "example_user", "content": "오늘은 날씨가 좋다"},
        {"role": "system", "name": "example_assistant", "content": "false"},
        {"role": "user", "content": "ChatGPT는 매우 편리하다"},
    ]
)
print(response.choices[0].message.content)
```

### Zero-shot Chain of Thought 프롬프팅

```python
response = client.chat.completions.create(
    model="gpt-6-luna",
    messages=[
        {"role": "system", "content": "답변만 한 마디로 출력해 주세요."},
        {"role": "user", "content": "10 + 2 * 3 - 4 * 2"}
    ],
    reasoning_effort="none"
)
print(response.choices[0].message.content)
```

```python
response = client.chat.completions.create(
    model="gpt-6-luna",
    messages=[
        {"role": "system", "content": "단계별로 생각해 보세요"},
        {"role": "user", "content": "10 + 2 * 3 - 4 * 2"}
    ],
    reasoning_effort="none"
)
print(response.choices[0].message.content)
```
