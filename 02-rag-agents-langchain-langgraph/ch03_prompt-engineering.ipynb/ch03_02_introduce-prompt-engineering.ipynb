```python
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv("../.env")

client = OpenAI()

response = client.chat.completions.create(
    model="gpt-6-luna",
    messages=[
        {"role": "user", "content": "프롬프트 엔지니어링이란"}
    ]
)
print(response.choices[0].message.content)
```

```python
responses = client.responses.create(
    model="gpt-6-luna",
    input="프롬프트 엔지니어링이란"
)
```

```python
print(responses.output_text)
```

```python
# 짧게 답변하도록 프롬프트를 개선해보겠다.
# Chat Completions API
response_2 = client.chat.completions.create(
    model="gpt-6-luna",
    messages=[
        {"role": "system", "content": "질문에 100자 정도로 답변해 주세요."},
        {"role": "user", "content": "프롬프트 엔지니어링이란"}
    ]
)
print(response_2.choices[0].message.content)
```

```python
# 짧게 답변하도록 프롬프트를 개선해보겠다.
# Responses API
responses_2 = client.responses.create(
    model="gpt-6-luna",
    instructions="질문에 100자 정도로 답변해 주세요.",
    input="프롬프트 엔지니어링이란"
)
print(responses_2.output_text)
```
